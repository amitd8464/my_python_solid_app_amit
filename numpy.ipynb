```python
import numpy as  np
from pprint import pprint


# Concatenation with axis
arr1 = np.array([[[1, 2], [3, 4]], [[1, 2], [3, 4]]])
arr2 = np.array([[[1, 2], [3, 4]], [[1, 2], [3, 4]]])
arr = np.concatenate((arr1, arr2), axis=1)

pprint(arr)

arr = np.concatenate((arr1, arr2), axis=0)

pprint(arr)
```
